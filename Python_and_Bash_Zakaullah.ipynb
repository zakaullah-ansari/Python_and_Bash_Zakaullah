### **1. Grade Checker**

Take a score as input and print the grade based on the following:


1. 90+ : “A”
2. 80–89 : “B”
3. 70–79 : “C”
4. 60–69 : “D”
5. Below 60 : “F”
6. Use a basic if / else statement to determine the grade.


```python
def check_grade(score):
    """Takes a score and returns the corresponding letter grade."""

    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"
```

```python
score = int(input("Enter the score:"))
check_grade(score)
```

### **2. Student Grades**

Create a dictionary where the keys are student names and the values are their grades. Allow the user to:

7. Add a new student and grade.
8. Update an existing student's grade.
9. Print all student grades.
10. Use a dictionary and basic operations with if / else.


```python
def manage_student_grades():
    """Demonstrates dictionary operations for adding, updating, and viewing student grades."""

    # Initial student grades dictionary
    grades_dict = {
        "Aamir": "A",
        "Raj": "B",
        "Arzoo": "C"
    }

    # 1. Print initial grades
    print("Initial Student Grades:", grades_dict)

    # 2. Add a new student
    new_name = input("Enter a name to add new student: ")
    new_grade = input(f"Enter a grade for {new_name}: ")
    grades_dict[new_name] = new_grade
    print(f"Added {new_name} with grade {new_grade}.")

    # Print updated dictionary
    print("Updated Student Grades:", grades_dict)

    # 3. Update an existing student's grade
    target_student = input("Enter the existing student's name: ")
    updated_grade = input(f"Enter the updated grade for {target_student}: ")
    if target_student in grades_dict:
        grades_dict[target_student] = updated_grade
        print(f"Updated {target_student}'s grade to {updated_grade}.")

    # 4. Print all student grades cleanly
    print("\nFinal Student Grades Roster:")
    for student, grade in grades_dict.items():
        print(f"  - {student}: {grade}")

manage_student_grades()
```

### **3. Write to a File**

11. Write a program to create a text file and write some content to it, using file functions like open() and write().

```python
def write_to_file(filename="assignment_output.txt"):
    """Creates a text file and writes sample content into it using open() and write()."""
    content = (
        "Hello, DevOps World!\n"
        "This file was created using Python file handling.\n"
        "Assignment 2 - Python & Bash Tasks Completed Successfully.\n"
    )

    with open(filename, "w") as file:
        file.write(content)

    print(f"Successfully wrote content to '{filename}'.")
```

### **4. Read from a File**

12. Open the file in read mode and use file.read() to read and display its content.

```python
import os

def read_from_file(filename="assignment_output.txt"):
    """Opens the text file in read mode and displays its content using file.read()."""

    if os.path.exists(filename):
        with open(filename, "r") as file:
            file_contents = file.read()
            print(f"Contents of '{filename}':")
            print("-" * 30)
            print(file_contents, end="")
            print("-" * 30)
    else:
        print(f"Error: The file '{filename}' does not exist.")


# ==========================================
# Main Execution Block
# ==========================================
if __name__ == "__main__":
    print("==========================================")
    print(" Running Assignment 2 Python Scripts")
    print("==========================================")

    # Test Task 1: Grade Checker
    print("\n--- Task 1: Grade Checker Test ---")
    test_scores = [95, 82, 74, 61, 45]
    for s in test_scores:
        print(f"Score {s} -> Grade: {check_grade(s)}")

    # Test Task 2: Student Grades Dictionary
    manage_student_grades()

    # Test Task 3: Write to File
    output_filename = "student_notes.txt"
    write_to_file(output_filename)

    # Test Task 4: Read from File
    read_from_file(output_filename)

    print("\nAll tasks executed successfully!")
```
